### List vs numpy array

```python
py_list = [1, 2, 3]
print("python list multiplication ", py_list * 2)

np_array = np.array([1, 2, 3]) #element wise multiplication
print("python array multiplication ", np_array *2)

import time
start = time.time()
py_list= [i*2 for i in range(1000000)]
print("\n List operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n Numpy operartion time: ", time.time() - start)
```

### Vector, Matrix and Tensor

```python
vector = np.array([1, 2, 3])
print("Vector: ", vector)

matrix = np.array([[1, 2, 3], 
                   [4, 5, 6]])
print("Matrix: ", matrix)

# it hae more dimension thats why its called tensor #tpu: tensor procesor unit

tensor = np.array([[[1,2], [3, 4]],
                   [[5, 6], [7, 8]]])
print("Tensor:", tensor) 
```

### Array Reshaping

```python
arr = np.arange(12)
print("Original array ", arr)

reshaped = arr.reshape((3,4))
print("\nReshaped array ", reshaped)

flattened = reshaped.flatten()
print("\nFlattened array ", flattened)

#asked in interviews too
# revel (returns view, instead of copy)
raveled = reshaped.ravel()
print("\n Raveled array ", raveled)

# Transpose
transpose = reshaped.T
print("Transposed array ", transpose)
```
