```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [resturant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
  [1, 150000, 180000, 220000, 250000], # Paradise Biryani
  [2, 150000, 180000, 220000, 250000], # Beijing Bites
  [3, 150000, 180000, 220000, 250000], # Pizza Hub
  [4, 150000, 180000, 220000, 250000], # Burger Point
  [5, 150000, 180000, 220000, 250000] # Chai Point
])

print("=== Zomato Sales Analysis")
print("\n Sales data shapes", sales_data.shape)
print("\n Sample data for first three restaurants:", sales_data[:, 1:]) 
```

```python
# total sales per year
print(np.sum(sales_data, axis=0))
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)

#Minimum sales per restaurants
min_sales = np.min(sales_data[:, 1:], axis=1)
print(min_sales)

#Maximum sales per year
max_sales = np.max(sales_data[:, 1], axis=0)
print(max_sales)

#Average sales per restaurant
avg_sales= np.mean(sales_data[:, 1:], axis=1)
print(avg_sales)

```

```python
#cummalative sales
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)

plt.Figure(figsize=(10, 6))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average cumulative sales per year accross all restaurant")
plt.xlabel("years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

### Vectorized Operation

```python
restaurant_types = np.array(['biriyani', 'chinese', 'pizza', 'burger', 'cafe'])
vectorized_upper = np.vectorize(str.upper)
print("Vectorized Upper", vectorized_upper(restaurant_types))
```
